```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix
```

```python
df["Embarked"].fillna("S",inplace=True)
```

```python
df["Cabin"].fillna("Unknown",inplace=True)
```

```python
df["Embarked"]=df["Embarked"].map({"S":0,"C":1,"Q":2})
```

```python
df['Sex']=df["Sex"].map({"male":0,"female":1})
```

```python
print(df.head())
```

```python
x=df.drop("Survived",axis=1)
y=df['Survived']
```

```python
x_train,x_test,y_train,y_test=train_test_split(
    x,y,test_size=0.2,random_state=42
)
```

```python
# Drop columns that are not suitable for numerical modeling
# Check if columns exist before dropping to prevent KeyError
columns_to_drop = ['PassengerId', 'Name', 'Ticket']
existing_columns_to_drop = [col for col in columns_to_drop if col in df.columns]
if existing_columns_to_drop:
    df.drop(existing_columns_to_drop, axis=1, inplace=True)

# Ensure 'Sex' and 'Embarked' are numerical and handle any NaNs.
# Based on the df.head() output, these columns might still contain NaNs
# after previous mapping attempts or due to inconsistencies in execution history.
# We will fill them with a sensible default if they still contain NaNs.
df['Sex'] = df['Sex'].fillna(0) # Assuming 0 for male is a reasonable default if NaN
df['Embarked'] = df['Embarked'].fillna(0) # Assuming 0 for 'S' is a reasonable default if NaN
```

```python
x = df.drop("Survived", axis=1)
y = df['Survived']
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)
```

```python
model=LogisticRegression(max_iter=1000)
# Ensure x_train contains only numerical features by explicitly dropping known non-numerical columns
columns_to_drop_from_x_train = ['Name', 'Ticket', 'Cabin', 'PassengerId']
x_train_cleaned = x_train.drop(columns=[col for col in columns_to_drop_from_x_train if col in x_train.columns], axis=1, errors='ignore')
model.fit(x_train_cleaned,y_train)
```

```python
y_pred = model.predict(x_test.drop(columns=columns_to_drop_from_x_train, errors='ignore'))

accuracy = accuracy_score(y_test, y_pred)
conf_matrix = confusion_matrix(y_test, y_pred)
sns.heatmap(conf_matrix,annot=True,fmt="d")
plt.title("Confusion Matrix")
plt.show()

print(f"Accuracy: {accuracy:.4f}")
print("Confusion Matrix:")
print(conf_matrix)
```

```python
#insight
#there were five people who were peredicted to be not survived and in real also they were not survived.
#There were 29 people who were actually survived and were predicted to be survived.
#There were 3 people actually died but it predicted they were not died.
#There were 4 people who were survived but model predicted that they were died.
#Here, x axis has predicted values and y axis has actual values in the confusion matrix.
```

```python
y_pred = model.predict(x_test.drop(columns=columns_to_drop_from_x_train, errors='ignore'))

accuracy = accuracy_score(y_test, y_pred)
conf_matrix = confusion_matrix(y_test, y_pred)

print(f"Accuracy: {accuracy:.4f}")
print("Confusion Matrix:")
print(conf_matrix)
```
